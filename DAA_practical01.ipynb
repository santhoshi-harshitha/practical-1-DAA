### Bubble Sort

```python

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

user_input = input("Enter numbers for Bubble Sort, separated by spaces: ")
numbers = list(map(int, user_input.split()))

sorted_numbers = bubble_sort(numbers)
print("Sorted array (Bubble Sort):", sorted_numbers)
```

#### Time Complexity (Bubble Sort):
-   **Best Case:** O(n) (already sorted)
-   **Average Case:** O(n^2)
-   **Worst Case:** O(n^2)

### Selection Sort

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

user_input = input("Enter numbers for Selection Sort, separated by spaces: ")
numbers = list(map(int, user_input.split()))

sorted_numbers = selection_sort(numbers)
print("Sorted array (Selection Sort):", sorted_numbers)
```

#### Time Complexity (Selection Sort):
-   **Best Case:** O(n^2)
-   **Average Case:** O(n^2)
-   **Worst Case:** O(n^2)

### Insertion Sort

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

user_input = input("Enter numbers for Insertion Sort, separated by spaces: ")
numbers = list(map(int, user_input.split()))

sorted_numbers = insertion_sort(numbers)
print("Sorted array (Insertion Sort):", sorted_numbers)
```

#### Time Complexity (Insertion Sort):
-   **Best Case:** O(n) (already sorted)
-   **Average Case:** O(n^2)
-   **Worst Case:** O(n^2)

### Merge Sort

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)

    return merge(left_half, right_half)

def merge(left, right):
    merged = []
    left_idx, right_idx = 0, 0

    while left_idx < len(left) and right_idx < len(right):
        if left[left_idx] < right[right_idx]:
            merged.append(left[left_idx])
            left_idx += 1
        else:
            merged.append(right[right_idx])
            right_idx += 1

    merged.extend(left[left_idx:])
    merged.extend(right[right_idx:])
    return merged

user_input = input("Enter numbers for Merge Sort, separated by spaces: ")
numbers = list(map(int, user_input.split()))

sorted_numbers = merge_sort(numbers)
print("Sorted array (Merge Sort):", sorted_numbers)
```

#### Time Complexity (Merge Sort):
-   **Best Case:** O(n log n)
-   **Average Case:** O(n log n)
-   **Worst Case:** O(n log n)

### Quick Sort

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

user_input = input("Enter numbers for Quick Sort, separated by spaces: ")
numbers = list(map(int, user_input.split()))

sorted_numbers = quick_sort(numbers)
print("Sorted array (Quick Sort):", sorted_numbers)
```

#### Time Complexity (Quick Sort):
-   **Best Case:** O(n log n)
-   **Average Case:** O(n log n)
-   **Worst Case:** O(n^2) (can be avoided with good pivot selection)
