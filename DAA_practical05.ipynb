```python
def knapsack_dp(weights, values, capacity):
    """
    Solves the knapsack problem using dynamic programming.

    Args:
        weights (list): A list of integers representing the weights of items.
        values (list): A list of integers representing the values of items.
        capacity (int): The maximum capacity of the knapsack.

    Returns:
        int: The maximum total value that can be achieved.
        list: A list of booleans indicating which items are included.
    """
    n = len(values)
    # dp[i][w] will store the maximum value that can be obtained with a knapsack
    # of capacity 'w' using items up to index 'i-1'.
    dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            # If the current item's weight is less than or equal to the current capacity
            if weights[i-1] <= w:
                # Option 1: Include the item. Value is current item's value + value from remaining capacity
                # Option 2: Exclude the item. Value is from the previous items with the same capacity
                dp[i][w] = max(values[i-1] + dp[i-1][w - weights[i-1]], dp[i-1][w])
            else:
                # If the item's weight is greater than current capacity, we cannot include it
                dp[i][w] = dp[i-1][w]

    # Reconstruct the chosen items
    selected_items = [False] * n
    w = capacity
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i-1][w]:
            selected_items[i-1] = True
            w -= weights[i-1]
```
