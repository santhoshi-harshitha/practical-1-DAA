```python
def linear_search(arr, target):
    """Performs a linear search to find the target in the array."""
    for i in range(len(arr)):
        if arr[i] == target:
            return i  # Return the index if target is found
    return -1 # Return -1 if target is not found
```

```python
def binary_search(arr, target):
    """Performs a binary search to find the target in a sorted array."""
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
```

```python
# Example Usage:
my_list = [4, 2, 7, 1, 9, 5]
target_element = 7

# Linear Search
linear_result = linear_search(my_list, target_element)
if linear_result != -1:
    print(f"Linear Search: Element {target_element} found at index {linear_result}")
else:
    print(f"Linear Search: Element {target_element} not found")

# Binary Search (requires a sorted list)
sorted_list = sorted(my_list)
binary_result = binary_search(sorted_list, target_element)
if binary_result != -1:
    print(f"Binary Search: Element {target_element} found at index {binary_result} in the sorted list {sorted_list}")
else:
    print(f"Binary Search: Element {target_element} not found in the sorted list {sorted_list}")


other_target = 3
linear_result_not_found = linear_search(my_list, other_target)
print(f"Linear Search: Element {other_target} not found (result: {linear_result_not_found})")

binary_result_not_found = binary_search(sorted_list, other_target)
print(f"Binary Search: Element {other_target} not found (result: {binary_result_not_found})")
```
